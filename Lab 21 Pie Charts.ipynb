### Task 1: Visualizing Data Composition with Pie Charts


##### 1.1 Create a Pie Chart of the Top 5 Databases Respondents Want to Work With


In the survey data, the `DatabaseWantToWorkWith` column lists the databases that respondents wish to work with. Let’s visualize the top 5 most-desired databases in a pie chart.



```python
##Write your code here
import pandas as pd
import matplotlib.pyplot as plt

# ---------------------------------------------------------
# STEP 1 — Clean and explode database preferences
# ---------------------------------------------------------
df['DatabaseWantToWorkWith'] = df['DatabaseWantToWorkWith'].fillna('')

df_db = df.assign(
    Database=df['DatabaseWantToWorkWith'].str.split(';')
).explode('Database')

df_db['Database'] = df_db['Database'].str.strip()
df_db = df_db[df_db['Database'] != '']

# ---------------------------------------------------------
# STEP 2 — Count database frequencies
# ---------------------------------------------------------
db_counts = df_db['Database'].value_counts().head(5)

# ---------------------------------------------------------
# STEP 3 — Pie chart
# ---------------------------------------------------------
plt.figure(figsize=(10,7))

plt.pie(
    db_counts,
    labels=db_counts.index,
    autopct='%1.1f%%',
    startangle=140
)

plt.title("Top 5 Databases Respondents Want to Work With")
plt.tight_layout()
plt.show()
```
