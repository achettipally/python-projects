Save into an excel spreadsheet named **job-postings.xlsx**.


```python
import requests
from openpyxl import Workbook


# API URL
api_url = "https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBM-DA0321EN-SkillsNetwork/labs/module%201/Accessing%20Data%20Using%20APIs/jobs.json"

# Fetch data
response = requests.get(api_url)
if response.ok:
    data = response.json()
else:
    raise Exception("Failed to fetch API data")

# Extract all unique locations
locations = set(job.get("Location", "").strip() for job in data)

# Count jobs per location
results = []
for loc in locations:
    count = sum(1 for job in data if job.get("Location", "").strip() == loc)
    results.append((loc, count))

# Create workbook
wb = Workbook()

# Select active worksheet
ws = wb.active
ws.title = "Job Counts by Location"

# Write header
ws.append(["Location", "Number of Jobs"])

# Write data rows
for loc, count in results:
    ws.append([loc, count])

# Save workbook
output_file = "job-postings.xlsx."
wb.save(output_file)

print(f"Workbook created and saved as: {output_file}")
```

#### In the similar way, you can try for below given key skills and results  can be stored in an excel sheet.


Collect the number of job postings for the following key skills using the API:

*   C
*   C#
*   C++
*   Java
*   JavaScript
*   Python
*   Scala
*   Oracle
*   SQL Server
*   MySQL Server
*   PostgreSQL
*   MongoDB


```python
# your code goes here
import requests
from openpyxl import Workbook

# API URL
api_url = "https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBM-DA0321EN-SkillsNetwork/labs/module%201/Accessing%20Data%20Using%20APIs/jobs.json"

# Fetch data
response = requests.get(api_url)
if response.ok:
    data = response.json()
else:
    raise Exception("Failed to fetch API data")

# List of key skills to search for
key_skills = [
    "C", "C#", "C++", "Java", "JavaScript", "Python",
    "Scala", "Oracle", "SQL Server", "MySQL Server",
    "PostgreSQL", "MongoDB"
]

# Function to count jobs for a given skill
def count_jobs_for_skill(skill):
    count = 0
    for job in data:
        skills = job.get("Key Skills", "").lower()
        if skill.lower() in skills:
            count += 1
    return count

# Create workbook
wb = Workbook()
ws = wb.active
ws.title = "Job Counts by Skill"

# Write header
ws.append(["Key Skill", "Number of Jobs"])

# Write data rows
for skill in key_skills:
    num_jobs = count_jobs_for_skill(skill)
    ws.append([skill, num_jobs])

# Save workbook
output_file = "job_counts_by_skill.xlsx"
wb.save(output_file)

print(f"Workbook created and saved as: {output_file}")

import pandas as pd
import matplotlib.pyplot as plt

# Load the Excel file you generated
df = pd.read_excel("job_counts_by_skill.xlsx")

# Sort by number of jobs in descending order
df_sorted = df.sort_values(by="Number of Jobs", ascending=False)

# Plot bar chart
plt.figure(figsize=(12,6))
plt.bar(df_sorted["Key Skill"], df_sorted["Number of Jobs"], color="steelblue")

# Labels and title
plt.xlabel("Key Skill")
plt.ylabel("Number of Job Postings")
plt.title("Job Postings by Skill (Descending Order)")

# Rotate x-axis labels for readability
plt.xticks(rotation=45, ha="right")

plt.tight_layout()
plt.show()


```
